```python
import os
import json

games = []

for file in os.listdir("games"):
    if file.endswith(".json"):
        with open(os.path.join("games", file)) as f:
            games.append(json.load(f))

games
```

```python
def retrieve_game_info(query):
    results = collection.query(
        query_texts=[query],
        n_results=1
    )
    return results

retrieve_game_info("Who developed FIFA 21?")
```
